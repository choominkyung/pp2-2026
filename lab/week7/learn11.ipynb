### lab_ 람다식으로 온도 변환하기

```python
def celsius(T):
    return (5.0/9.0)*(T-32.0)

f_temp = [0,10,20,30,40,50]
c_temp = map(celsius, f_temp)
print(list(c_temp))
```

### lab_ 피보나치 이터레이터


```python
class FibIterator:
    def __init__(self, a=1, b=0, maxValue =50):
        self.a =a
        self.b=b
        self.maxValue=maxValue
    def __iter__(self):
        return self
    def __next__(self):
        n=self.a+self.b
        if n> self.maxValue:
            raise StopIteration()
        self.a = self.b
        self.b = n
        return n

for i in FibIterator():
    print(i, end =" ")
```

### lab_ Book클래스

```python
class Book:
    title = ''
    pages = 0

    def __init__(self, title='', pages=0):
        self.title = title
        self.pages = pages

    def __str__(self):
        return self.title

    def __gt__(self, other):
        return self.pages > other.pages


book1 = Book('Magic of Python', 600)
book2 = Book('Master of Python', 700)

print(book1 > book2)
```

### lab_ 동전던지기 게임
