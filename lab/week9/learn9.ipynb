# 9장 GUI 프로그래밍

- tkinter란?  
: 파이썬에서 그래픽 사용자 인터페이스를 개발 할 때 필요한 모듈이다. tkinter를 사용하면 윈도우를 생성하고 버튼이나 레이블과 같은 위젯을 이용하여서 사용자와 상호작용하는 프로그램을 작성할 수 있다. turtle을 예전에 사용했었지만 속도가 느리므로 마우스와 키보드에 입력받을 수 있는tkinter에 대해서 배워보는 9장내용이다.

tkinter 시작하기  


 윈도우 생성하고 여기에 필요한 위젯  추가하기  
여기서 위젯이란? GUI 기반 시스템에서 사용하는 각종 시각적인 요소를 말함. 예를들어 버튼, 레이블, 슬라이더, 콤보 박스 등이다  



```python
from tkinter import * #tkinter모듈을 포함
window=Tk() # 루트 윈도우를 생성
label= Label(window, text="Hello tkinter")#레이블 위젯을 생성
label.pack()#레이블 위젯을 윈도우에 배치

window.mainloop()#윈도우가 사용자 동작을 대기
```

1. from tkinter import   
: tkinter모듈을 포함시키는 것 tkinter이라는 모듈에서 *모든 것을 포함하라는 것이다.  
2. window=Tk()   
: 루트 윈도우를 생성하여야한다. 여러가지 위젯을 추가할 수 있다.  
3. label= Label(window, text="Hello tkinter")  
: Label의 첫 번째 매개변수는 항상 부모 컨테이너이다. 컨테이너란 위젯을 내부에 가질 수 있는 객체이다. 두번째 매개변수는 레이블에 표시되는 텍스트이다.   
4. label.pack()  
: 압축 배치 관리자를 이용해서 레이블 컨테이너에  위치시킨다.  
5. window.mainloop()   
: 맨끝에 이문장을 가져야 한다.  

기본위젯들

- Label:  
텍스트를 표시하는데 사용하는 위젯  
- Butten:  
버튼을 제공하는 위젯
-Entry:  
한줄의 텍스트를 입력받는 위젯


배치관리자

-Pack  
:압축 배치 관리자, 사각형 블록으로 간주하여서 상하로 배치한다.  
- Grid  
: 격자배치 관리자. 위젯들을 2차적인 격자에 배치한다.   
- Place  
: 절대 배치 관리자, 사용자가 지정한 위치에 위젯들을 배치한다.

### lab_숫자추측게임

```python
from tkinter import *
import random

answer = random.randint(1, 100)  # 정답을 1에서 100 사이의 난수로 설정한다.

def guessing():
    guess = int(guessField.get())  # 텍스트 필드에서 사용자가 입력한 값을 가져온다.

    if guess > answer:
        msg = "높음!"
    elif guess < answer:
        msg = "낮음!"
    else:
        msg = "정답!"

    resultLabel["text"] = msg  # 메시지를 출력한다.
    guessField.delete(0, 5)

def reset():
    global answer  # 정답을 다시 설정한다.
    answer = random.randint(1, 100)
    resultLabel["text"] = "다시 한번 하세요!"

window = Tk()
window.configure(bg="white")
window.title("숫자를 맞춰보세요!")

window.geometry("500x80")

titleLabel = Label(window, text="숫자 게임에 오신 것을 환영합니다!", bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side="left")

tryButton = Button(window, text="시도", fg="green", bg="white", command=guessing)
tryButton.pack(side="left")

resetButton = Button(window, text="초기화", fg="red", bg="white", command=reset)
resetButton.pack(side="left")

resultLabel = Label(window, text="1부터 100사이의 숫자를 입력하시오.", bg="white")
resultLabel.pack(side="left")

window.mainloop()
```

### lab_ 가위바위보게임

```python
##
# 이 프로그램은 가위, 바위, 보 게임을 구현한다.[span_2](start_span)[span_2](end_span)
#
import random
from tkinter import *

window = Tk()
Label(window, text="선택하세요", font=("Helvetica", "16")).pack()
frame = Frame(window)[span_4](start_span)[span_4](end_span)

# 이미지를 읽어서 객체로 만든다.[span_5](start_span)[span_5](end_span)
rock_image = PhotoImage(file="rock.png")
paper_image = PhotoImage(file="paper.png")
scissors_image = PhotoImage(file="scissors.png")

def pass_s():
    decide("가위")[span_6](start_span)[span_6](end_span)

def pass_r():
    decide("바위")[span_7](start_span)[span_7](end_span)

def pass_p():
    decide("보")[span_8](start_span)[span_8](end_span)

# 이미지 버튼들을 생성한다.[span_9](start_span)[span_9](end_span)
rock = Button(frame, image=rock_image, command=pass_r)[span_10](start_span)[span_10](end_span)
rock.pack(side="left")[span_11](start_span)[span_11](end_span)
paper = Button(frame, image=paper_image, command=pass_p)[span_12](start_span)[span_12](end_span)
paper.pack(side="left")[span_13](start_span)[span_13](end_span)
scissors = Button(
    frame, image=scissors_image, command=pass_s
)[span_14](start_span)[span_14](end_span)  # 이미지 버튼 생성
scissors.pack(side="left")[span_15](start_span)[span_15](end_span)

frame.pack()[span_16](start_span)[span_16](end_span)

# 정보를 표시하는 레이블을 생성한다.[span_17](start_span)[span_17](end_span)
Label(window, text="컴퓨터는 다음을 선택하였습니다.", font=("Helvetica", "16")).pack()[
    cite: 11
]
computer_image = Label(window, image=rock_image)[span_18](start_span)[span_18](end_span)
computer_image.pack()[span_19](start_span)[span_19](end_span)
output = Label(window, text="", font=("Helvetica", "16"))[span_20](start_span)[span_20](end_span)
output.pack()[span_21](start_span)[span_21](end_span)

def decide(human):
    computer = random.choice(["가위", "바위", "보"])[span_22](start_span)[span_22](end_span)
    if computer == "바위":
        computer_image["image"] = rock_image[span_23](start_span)[span_23](end_span)
    elif computer == "보":
        computer_image["image"] = paper_image[span_24](start_span)[span_24](end_span)
    else:
        computer_image["image"] = scissors_image[span_25](start_span)[span_25](end_span)

    # 승리 또는 패배를 판단한다.[span_26](start_span)[span_26](end_span)
    if (
        (computer == "바위" and human == "보")
        or (computer == "보" and human == "가위")
        or (computer == "가위" and human == "바위")
    ):
        result = "인간 승리[span_27](start_span)!"[span_27](end_span)
    elif computer == human:
        result = "비겼습니다.[span_28](start_span)"[span_28](end_span)
    else:
        result = "컴퓨터 승리[span_29](start_span)!"[span_29](end_span)
    output.config(text="인간: " + human + " 컴퓨터: " + computer + "  " + result)[
        cite: 11
    ]

window.mainloop()[span_30](start_span)[span_30](end_span)
```

### lab_ tictactoe


```python
from tkinter import *

# i번째 버튼을 누를 수 있는지 검사한다. 누를 수 있으면 X나 O를 표시한다.
def checked(i):
    global player
    button = list[i] # 리스트에서 i번째 버튼 객체를 가져온다.

    # 버튼이 초기상태가 아니면 이미 누른 버튼이므로 아무것도 하지 않고 리턴한다.
    if button["text"] != "  ":
        return
    
    button["text"] = "  " + player + "  "
    
    if player == "X":
        player = "O"
        button["bg"] = "yellow"
    else:
        player = "X"
        button["bg"] = "lightgreen"

window = Tk()             # 윈도우를 생성한다.
player = "X"              # 시작은 플레이어 X이다.
list = []

# 9개의 버튼을 생성하여 격자 형태로 윈도우에 배치한다.
for i in range(9):
    b = Button(window, text="  ", command=lambda k=i: checked(k))
    b.grid(row=i//3, column=i%3)
    list.append(b)       # 버튼 객체를 리스트에 저장한다.

window.mainloop()
```

### lab_계산기 프로그램

```python
from tkinter import *

window = Tk()
window.title("My Calculator")
display = Entry(window, width=33, bg="yellow")
display.grid(row=0, column=0, columnspan=5)

button_list = [
    '7', '8', '9', '/', 'C',
    '4', '5', '6', '*', '(',
    '1', '2', '3', '-', ')',
    '0', '.', '=', '+', ' '
]

def click(key):
    if key == "=":
        result = eval(display.get())
        s = str(result)
        display.insert(END, "=" + s)
    else:
        display.insert(END, key)

row_index = 1
col_index = 0

for button_text in button_list:
    Button(window, text=button_text, width=5,
           command=lambda t=button_text: click(t)).grid(row=row_index,
           column=col_index)
    col_index += 1
    if col_index > 4:
        row_index += 1
        col_index = 0

window.mainloop()
```

### lab_도형그리기

```python
from tkinter import *

WIDTH = 600
HEIGHT = 200

def displayRect():
    canvas.create_rectangle(10, 10, WIDTH-10, HEIGHT-10)

def displayOval():
    canvas.create_oval(10, 10, WIDTH-10, HEIGHT-10, fill="yellow")

def displayArc():
    canvas.create_arc(10, 10, WIDTH-10, HEIGHT-10, start=0,
                      extent=120, width=10, fill='blue')

def displayPolygon():
    canvas.create_polygon(10, 10, WIDTH-10, HEIGHT-10, 200, 90, 300, 160)

def displayLine():
    canvas.create_line(10, 10, WIDTH-10, HEIGHT-10, fill='green')

def clearCanvas():
    canvas.delete(ALL)

window = Tk()
canvas = Canvas(window, width=WIDTH, height=HEIGHT, bg='white')
canvas.pack()
frame = Frame(window)
frame.pack()

btRectangle = Button(frame, text="Rectangle", command=displayRect)
btRectangle.grid(row=1, column=2)
btOval = Button(frame, text="Oval", command=displayOval)
btOval.grid(row=1, column=3)
btArc = Button(frame, text="Arc", command=displayArc)
btArc.grid(row=1, column=5)
btPolygon = Button(frame, text="Polygon", command=displayPolygon)
btPolygon.grid(row=1, column=4)
btLine = Button(frame, text="Line", command=displayLine)
btLine.grid(row=1, column=1)
btClear = Button(frame, text="Clear", command=clearCanvas)
btClear.grid(row=1, column=7)

window.mainloop()
```

### lab_그림판프로그램만들기

```python
from tkinter import *
from tkinter.colorchooser import askcolor

DEFAULT_PEN_SIZE = 1.0
DEFAULT_COLOR = "black"

mode = "pen"
old_x = None
old_y = None
mycolor = DEFAULT_COLOR
erase_on = False

def use_pen():          # 펜 버튼이 선택되면 모드를 "pen"으로 바꾼다.
    global mode
    mode = "pen"

def use_brush():        # 브러쉬 버튼이 선택되면 모드를 "brush"로 바꾼다.
    global mode
    mode = "brush"

def choose_color():     # 색상 버튼이 선택되면 사용자한테 색상을 요구한다.
    global mycolor
    mycolor = askcolor(color=mycolor)[1]

def use_erase():        # 지우개 버튼이 선택되면 모드를 "erase"로 바꾼다.
    global mode
    mode = "erase"

def paint(event):       # 이전 점과 현재 점 사이를 직선으로 연결한다.
    global var, erase_on, mode, old_x, old_y
    fill_color = 'white' if mode == 'erase' else mycolor
    if old_x and old_y:
        canvas.create_line(old_x, old_y, event.x, event.y,
                           capstyle=ROUND, width=var.get(), fill=fill_color)
    old_x = event.x
    old_y = event.y

def reset(event):       # 사용자가 마우스 버튼에서 손을 떼면 이전 점을 삭제한다.
    global old_x, old_y
    old_x, old_y = None, None

def clear_all():        # 캔버스에 그려진 모든 그림을 삭제한다.
    global canvas
    canvas.delete('all')

window = Tk()           # 슬라이더와 연결될 객체를 생성한다.
var = DoubleVar()
penButton = Button(window, text='펜', command=use_pen)
penButton.grid(row=0, column=0, sticky=W+E)

brushButton = Button(window, text='브러쉬', command=use_brush)
brushButton.grid(row=0, column=1, sticky=W+E)

colorButton = Button(window, text='색상선택', command=choose_color)
colorButton.grid(row=0, column=2, sticky=W+E)

eraseButton = Button(window, text='지우개', command=use_erase)
eraseButton.grid(row=0, column=3, sticky=W+E)

clearButton = Button(window, text='모두삭제', command=clear_all)
clearButton.grid(row=0, column=4, sticky=W+E)

scale = Scale(window, variable=var, orient=VERTICAL)
scale.grid(row=1, column=5, sticky=N+S)

canvas = Canvas(window, bg='white', width=600, height=400)
canvas.grid(row=1, columnspan=5)

canvas.bind('<B1-Motion>', paint)
canvas.bind('<ButtonRelease-1>', reset)

window.mainloop()
```

### lab_공 애니메이션

```python
from tkinter import *
import time
import random

window = Tk()
canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball():
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        if y1 > canvas.winfo_height() or y0 < 0:  # 원이 위쪽이나 아래쪽으로 벗어났으면
            self.dy = -self.dy                    # dy의 부호를 반전시킨다.
        if x1 > canvas.winfo_width() or x0 < 0:   # 원이 왼쪽이나 오른쪽으로 벗어났으면
            self.dx = -self.dx                    # dx의 부호를 반전시킨다.

ball1 = Ball("blue", 60)
ball2 = Ball("green", 100)
ball3 = Ball("orange", 80)

while True:
    ball1.move()
    ball2.move()
    ball3.move()
    window.update()
    time.sleep(0.05)

window.mainloop()
```
