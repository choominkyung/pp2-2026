# 9장 GUI 프로그래밍

- tkinter란?  
: 파이썬에서 그래픽 사용자 인터페이스를 개발 할 때 필요한 모듈이다. tkinter를 사용하면 윈도우를 생성하고 버튼이나 레이블과 같은 위젯을 이용하여서 사용자와 상호작용하는 프로그램을 작성할 수 있다. turtle을 예전에 사용했었지만 속도가 느리므로 마우스와 키보드에 입력받을 수 있는tkinter에 대해서 배워보는 9장내용이다.

tkinter 시작하기  


 윈도우 생성하고 여기에 필요한 위젯  추가하기  
여기서 위젯이란? GUI 기반 시스템에서 사용하는 각종 시각적인 요소를 말함. 예를들어 버튼, 레이블, 슬라이더, 콤보 박스 등이다  



```python
from tkinter import * #tkinter모듈을 포함
window=Tk() # 루트 윈도우를 생성
label= Label(window, text="Hello tkinter")#레이블 위젯을 생성
label.pack()#레이블 위젯을 윈도우에 배치

window.mainloop()#윈도우가 사용자 동작을 대기
```

1. from tkinter import   
: tkinter모듈을 포함시키는 것 tkinter이라는 모듈에서 *모든 것을 포함하라는 것이다.  
2. window=Tk()   
: 루트 윈도우를 생성하여야한다. 여러가지 위젯을 추가할 수 있다.  
3. label= Label(window, text="Hello tkinter")  
: Label의 첫 번째 매개변수는 항상 부모 컨테이너이다. 컨테이너란 위젯을 내부에 가질 수 있는 객체이다. 두번째 매개변수는 레이블에 표시되는 텍스트이다.   
4. label.pack()  
: 압축 배치 관리자를 이용해서 레이블 컨테이너에  위치시킨다.  
5. window.mainloop()   
: 맨끝에 이문장을 가져야 한다.  

기본위젯들

- Label:  
텍스트를 표시하는데 사용하는 위젯  
- Butten:  
버튼을 제공하는 위젯
-Entry:  
한줄의 텍스트를 입력받는 위젯


배치관리자

-Pack  
:압축 배치 관리자, 사각형 블록으로 간주하여서 상하로 배치한다.  
- Grid  
: 격자배치 관리자. 위젯들을 2차적인 격자에 배치한다.   
- Place  
: 절대 배치 관리자, 사용자가 지정한 위치에 위젯들을 배치한다. 

### lab_온도 변환기 

```python
from tkinter import *

window = Tk()

Label(window, text="화씨").grid(row=0, column=0)
Label(window, text="섭씨").grid(row=1, column=0)

e1 = Entry(window).grid(row=0, column=1)
e2 = Entry(window).grid(row=1, column=1)

Button(window, text="화씨->섭씨").grid(row=2, column=1)

window.mainloop()
```
