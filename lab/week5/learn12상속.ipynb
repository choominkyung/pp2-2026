```python

class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery: {self.batterySize}kwh"

    def getDesc(self):
        return self.__str__()


class Truck(Car):
    def __init__(self, maker, model, color, price, payload):
        super().__init__(maker, model, color, price)
        self.payload = payload

    def getPayload(self):
        return self.payload

    def setPayload(self, payload):
        self.payload = payload

    def __str__(self):
        return f"{super().__str__()}, Payload: {self.payload}kg"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)


ecar1= ElectricCar("Tesla", "Model S", "Red", 8000,81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc()) 
truck1 = Truck("Ford", "F-150", "Black", 4000,1000)
print(truck1)
print(truck1.getMaker())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

       
```

# 속성변수의 가시성 : public, protected(_로 시작)vate(_로 시작)

```python
class Parent(object):
    def __init__(self):
        self._private_money = 1000
        self.protected_money = 500
        self.public_money = 100

    def __str__(self):
        return f"Parent has total {self._private_money + self.protected_money + self.public_money} money, private money: {self._private_money}, protected money: {self.protected_money}, public money: {self.public_money}"


class Child(Parent):
    def __init__(self):
        super().__init__()

    def __str__(self):
        return f"Child has total {self._private_money + self.protected_money + self.public_money} money, private money: {self._private_money}, protected money: {self.protected_money}, public money: {self.public_money}"


def test():
    p = Parent()
    print(p)

    c = Child()
    print(c)

    print(p.public_money)
    print(p.protected_money)

    print(c.public_money)
    print(c.protected_money)

    

test()
```

# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1=10):
        self.attr1 = v1
    def display1(self):
        print(f"{self.attr1}")


class Base2:
    def __init__(self, v2=20):
        self.attr2 = v2
    def display2(self):
        print(f"{self.attr2}")



class C(Base1, Base2):
    def __init__(self, v1,v2,v3):
        Base1.__init__(self,v1)
        Base2.__init__(self,v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
       


def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2=C(4,5,6)
    c2.print()

if __name__ == "__main__":
    test()
```

# Overriding :부모의 함수(메소드)재정의 

```python
import math


class Shape:
    def __init__(self, name="도형"):
        self.name = name

    def draw(self):
        print(f"{self.name}을(를) 그립니다.")

    def get_area(self):
        print(f"Shape: {self.name}의 면적을 계산합니다")


class Circle(Shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius

    def draw(self):
        print(f"Circle: {self.name}을(를) 그립니다.")

    def get_area(self) -> float:
        print(f"Circle: {self.name}의 면적을 계산합니다.")
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width =1, height=1, name ="사각형"):
        super().__init__(name)
        self.width =width
        self.height = height
    def draw(self):
            print(f"Rectangle:{self.name}을(를)그립니다.")
    
    def get_area(self) -> float:
            print(f"Rectangle:{self.name}의 면적을 계산합니다.")
            return self.width *self.height

def test():
    c = Circle(10)
    c.draw()
    print(c.get_area())

    r1= Rectangle(width=5, height =10)
    r1.draw()
    print (r1.get_area())
    
    print("-----------------------------------")
    circles=[Circle(1), Circle(2), Circle(3)]
    for c in circles:
         c.draw()
         print(c.get_area())
       
 



if __name__ == "__main__":
    test()
```

```python
##
# 이 프로그램은 메서드 오버라이딩을 구현한다.
#

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def getSalary(self):
        return self.salary

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def getSalary(self):
        salary = super().getSalary()
        return salary + self.bonus

    def __str__(self):
        return "이름: " + self.name + "; 월급: " + str(self.salary) + \
               "; 보너स: " + str(self.bonus)

kim = Manager("김철수", 2000000, 1000000)
print(kim)
```

### lab_ Card와 Deck

```python
##
# 클래스 간의 HAS-A 관계를 구현한다.
#
class Card:
    suitNames = ['클럽', '다이아몬드', '하트', '스페이드']
    rankNames = [None, '에이스', '2', '3', '4', '5', '6', '7',
                 '8', '9', '10', '잭', '퀸', '킹']

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return Card.suitNames[self.suit] + " " + \
               Card.rankNames[self.rank]

class Deck:
    def __init__(self):
        self.cards = []
        for suit in range(4):
            for rank in range(1, 14):
                card = Card(suit, rank)
                self.cards.append(card)

    def __str__(self):
        lst = [str(card) for card in self.cards]
        return str(lst)

deck = Deck()             # 덱 객체를 생성한다.
print(deck)               # 덱 객체를 출력한다. __str__()이 호출된다.
```

### lab_ 학생과 강사

```python
class Person:
    def __init__(self, name, number):
        self.name = name
        self.number = number

class Student(Person):
    UNDERGRADUATE = 0
    POSTGRADUATE = 1

    def __init__(self, name, number, studentType):
        super().__init__(name, number)
        self.studentType = studentType
        self.gpa = 0
        self.classes = []

    def enrollCourse(self, course):
        self.classes.append(course)

    def __str__(self):
        return "\n이름=" + self.name + "\n주민번호=" + self.number + \
               "\n수강과목=" + str(self.classes) + "\n평점=" + str(self.gpa)

class Teacher(Person):
    def __init__(self, name, number):
        super().__init__(name, number)
        self.courses = []
        self.salary = 3000000

    def assignTeaching(self, course):
        self.courses.append(course)

    def __str__(self):
        return "\n이름=" + self.name + "\n주민번호=" + self.number + \
               "\n강의과목=" + str(self.courses) + "\n월급=" + str(self.salary)

hong = Student("홍길동", "12345678", Student.UNDERGRADUATE)
hong.enrollCourse("자료구조")
print(hong)

kim = Teacher("김철수", "123456790")
kim.assignTeaching("Python")
print(kim)
```

### lab_tkinter프로그램에서 상속

```python
from tkinter import *
import random

class Car(Canvas):
    def __init__(self, master = None):
        Canvas.__init__(self, master, width = 500, height = 100)
        self.master = master
        self.dx = 0
        self.dy = 0
        self.rectangle = self.create_rectangle(0, 5, 70, 35, fill = "black")
        self.oval = self.create_oval(5, 20, 25, 50, fill = "black")
        self.oval2 = self.create_oval(40, 20, 60, 50, fill = "black")
        self.pack()
        self.movement()

    def movement(self):
        dx = random.randint(2, 10)
        self.move(self.rectangle, dx, 0)
        self.move(self.oval, dx, 0)
        self.move(self.oval2, dx, 0)
        self.after(100, self.movement)

master = Tk()
car1 = Car(master)
car2 = Car(master)

mainloop()
```
