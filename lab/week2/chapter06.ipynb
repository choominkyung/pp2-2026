### lab_ 성적 처리 프로그램


```python
STUDENTS = 5
lst = []
count = 0

for i in range(STUDENTS):
    value = int(input("성적을 입력하시오: "))
    lst.append(value)

print("\n성적 평균=", sum(lst) / len(lst))
print("최대점수=", max(lst))
print("최소점수=", min(lst))

for score in lst:
    if score >= 80:
        count += 1
print("80점 이상=", count)
```

### lab_ 리스트 변경 방법 비교

```python
import time
SIZE = 50000

start_time = time.time()
mylist = []
for i in range(SIZE):
    mylist = mylist + [i * i]
print("수행시간=", time.time() - start_time)

start_time = time.time()
mylist = []
for i in range(SIZE):
    mylist.append(i * i)
print("수행시간=", time.time() - start_time)
```

### lab_ 리스트 변경함수

```python
salaries = [200, 250, 300, 280, 500]

def modify(values, factor) :
    for i in range(len(values)) :
        values[i] = values[i] * factor

print("인상전", salaries)
modify(salaries, 1.3)
print("인상후", salaries)
```

### lab_ 피타고라스삼각형

```python
new_list = []
for x in range(1, 30):
    for y in range(x, 30):
        for z in range(y, 30):
            if x**2 + y**2 == z**2:
                new_list.append((x, y, z))
print(new_list)
```

### lab_전치행렬계산


```python
transposed = []
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print("원래 행렬=", matrix)
# 열의 개수만큼 반복한다.
for i in range(len(matrix[0])):
    transposed_row = []
    for row in matrix:
        transposed_row.append(row[i])
    transposed.append(transposed_row)

print("전치 행렬=", transposed)
```

### lab_  tic tac toe게임

```python
board= [[' ' for x in range(3)] for y in range(3)]
while True:
    # 게임 보드를 그린다.
    for r in range(3):
        print("  " + board[r][0] + " | " + board[r][1] + " | "
              + board[r][2])
        if (r != 2):
            print("---|---|---")

    # 사용자로부터 좌표를 입력받는다.
    x = int(input("다음 수의 x좌표를 입력하시오: "))
    y = int(input("다음 수의 y좌표를 입력하시오: "))

    # 사용자가 입력한 좌표를 검사한다.
    if board[x][y] != ' ':
        print("잘못된 위치입니다. ")
        continue
    else:
        board[x][y] = 'X'

    # 컴퓨터가 놓을 위치를 결정한다. 첫 번째로 발견하는 비어있는 칸에 놓는다.
    done = False
    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ' and not done:
                board[i][j] = 'O'
                done=True
                break
```
