##깊은복사 _재귀적으로 리스트 안의 소속 객체들을 복사: copy

```python
import copy

temps= [[1,2],3,4,5]
svalues=copy.copy(temps)
temps[0][0]=100
temps
svalues
dvalues=copy.deepcopy(temps)
temps[0][0]=7
temps
dvalues

```
