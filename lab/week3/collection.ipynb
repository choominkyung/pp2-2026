
#### Python의 byte, bytes, bytearray

파이썬에서 `bytes`와 `bytearray`는 **문자열이나 데이터를 바이트(byte) 단위로 다루기 위한 자료형**이다. 주로 파일, 이미지, 네트워크 통신, 인코딩된 문자열처럼 컴퓨터가 실제로 저장하고 전송하는 데이터를 처리할 때 사용한다.

##### 1. Byte란?

**Byte(바이트)**는 컴퓨터에서 데이터를 저장하는 기본적인 단위 중 하나이다.

1 Byte는 **8개의 bit(비트)**로 이루어져 있으며, 0부터 255까지 총 256개의 값을 표현할 수 있다.

예를 들어 숫자 `65`는 하나의 바이트로 표현할 수 있다.

```python
65
```

문자 `'A'`도 컴퓨터 내부에서는 특정한 바이트 값으로 표현할 수 있다. ASCII에서는 `'A'`가 숫자 `65`에 해당한다.





* bit → 0 또는 1 하나
* byte → 8개의 bit
* bytes → 여러 개의 바이트로 이루어진 데이터



##### 2. bytes란?

`bytes`는 **바이트들의 묶음을 저장하는 파이썬 자료형**이다.

문자열을 `bytes`로 변환하면 문자 자체가 아니라 컴퓨터가 사용하는 바이트 형태로 저장된다.
python
text = "Hello"
data = text.encode()

print(data)


결과:

python
b'Hello'


앞에 붙어 있는 `b`는 이 데이터가 일반적인 문자열(`str`)이 아니라 **bytes 자료형**이라는 것을 의미한다.

자료형을 확인할 수도 있다.

```python
print(type(data))
```

결과:

```python
<class 'bytes'>
```

##### bytes의 특징

`bytes`는 **변경할 수 없는(immutable) 자료형**이다.
 `bytes`는 만들어진 후 내부의 바이트 값을 직접 수정할 수 없다. 새로운 bytes를 만들어야 한다.





##### 3. bytearray란?

`bytearray`도 여러 개의 바이트를 저장하는 자료형이다.

`bytes`와 거의 비슷하지만 가장 중요한 차이는 **값을 변경할 수 있다는 것**이다.



* `bytes` → 변경 불가능
* `bytearray` → 변경 가능
