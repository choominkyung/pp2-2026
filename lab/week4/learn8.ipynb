```python
from datetime import datetime
class person:

    def __init__(self, name, birth_year):
        self.__name= name
        self.__birth_year =birth_year

    def get_name(self):
        return self.__name

    def get_age(self):
        current_year = datetime.now().year
        return current_year - self.__birth_year
    
    def get_current_year(self):
        return datetime.now().year
    
    def __str__(self):
        return f"Name: {self.__name}, Age: {self.get_age()}"
    
    def test_person():
        song=person("Song", 2001)
        y=song.get_current_year()
        print(y)
        team_members=[person("Kim",2000), person("Lee", 1999)]
        print(team_members)

    if __name__ == "__main__":
        test_person()
```

### lab _ 원 클래스작성

```python
import math

# Circle 클래스를 정의한다.
class Circle:
    def __init__(self, radius = 0):
        self.radius = radius

    def getArea(self):
        return math.pi * self.radius * self.radius

    def getPerimeter(self):
        return 2 * math.pi * self.radius

# Circle 객체를 생성한다.
c = Circle(10)
print("원의 면적", c.getArea())
print("원의 둘레", c.getPerimeter())
```

### lab _ 은행 계좌

```python
##
# 이 프로그램은 은행 계좌를 클래스로 정의한다.
#
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def withdraw(self, amount):
        self.__balance -= amount
        print("통장에 ", amount, "가 입금되었음")
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount
        print("통장에서 ", amount, "가 출금되었음")
        return self.__balance

a = BankAccount()
a.deposit(100)
a.withdraw(10)
```

### lab_ 백터 객체에 특수 메소드 정의하기

```python
class Vector2D:

    def __init__(self, x, y):
        self.x = x  # x 좌표 저장
        self.y = y  # y 좌표 저장

    def __add__(self, other):
        # + 연산자 Overloading
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        # - 연산자 Overloading
        return Vector2D(self.x - other.x, self.y - other.y)

    def __eq__(self, other):
        # == 연산자 Overloading
        return self.x == other.x and self.y == other.y

    def __str__(self):
        # print() 또는 str() 호출 시 문자열 표현 정의
        return "(%g, %g)" % (self.x, self.y)


# 객체 생성 및 테스트
u = Vector2D(0, 1)
v = Vector2D(1, 0)
w = Vector2D(1, 1)

a = u + v  # Vector2D(0+1, 1+0) -> Vector2D(1, 1) 생성
print(a)  # __str__ 메소드가 호출되어 (1, 1) 출력
```
