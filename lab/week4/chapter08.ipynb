### lab _ 원 클래스작성

```python
import math

# Circle 클래스를 정의한다.
class Circle:
    def __init__(self, radius = 0):
        self.radius = radius

    def getArea(self):
        return math.pi * self.radius * self.radius

    def getPerimeter(self):
        return 2 * math.pi * self.radius

# Circle 객체를 생성한다.
c = Circle(10)
print("원의 면적", c.getArea())
print("원의 둘레", c.getPerimeter())
```

### lab _ 은행 계좌

```python
##
# 이 프로그램은 은행 계좌를 클래스로 정의한다.
#
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def withdraw(self, amount):
        self.__balance -= amount
        print("통장에 ", amount, "가 입금되었음")
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount
        print("통장에서 ", amount, "가 출금되었음")
        return self.__balance

a = BankAccount()
a.deposit(100)
a.withdraw(10)
```

### lab_ 백터 객체에 특수 메소드 정의하기

```python
class Vector2D:

    def __init__(self, x, y):
        self.x = x  # x 좌표 저장
        self.y = y  # y 좌표 저장

    def __add__(self, other):
        # + 연산자 Overloading
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        # - 연산자 Overloading
        return Vector2D(self.x - other.x, self.y - other.y)

    def __eq__(self, other):
        # == 연산자 Overloading
        return self.x == other.x and self.y == other.y

    def __str__(self):
        # print() 또는 str() 호출 시 문자열 표현 정의
        return "(%g, %g)" % (self.x, self.y)


# 객체 생성 및 테스트
u = Vector2D(0, 1)
v = Vector2D(1, 0)
w = Vector2D(1, 1)

a = u + v  # Vector2D(0+1, 1+0) -> Vector2D(1, 1) 생성
print(a)  # __str__ 메소드가 호출되어 (1, 1) 출력
```

```python
class Rocket:

    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return f"로켓의 위치: ({self.x}, {self.y})"

    def moveUp(self):
        self.y += 1


myRocket = Rocket()
print("로켓의 높이:", myRocket.y)

myRocket.moveUp()
print("로켓의 높이:", myRocket.y)
```

### programming 03  
상자를 나타내는  Box 클래스를 작성하여 보자. Box 클래스는 가로길이, 세로길이, 높이를 나타내는 인스턴스 변수를 가진다. 

```python
class Box:

    def __init__(self, l, h, d):
        self.__length = l
        self.__height = h
        self.__depth = d

    def __str__(self):
        return f"({self.__length}, {self.__height}, {self.__depth})"

    def getLength(self):
        return self.__length

    def setLength(self, length):
        self.__length = length

    def getHeight(self):
        return self.__height

    def setHeight(self, height):
        self.__height = height

    def getDepth(self):
        return self.__depth

    def setDepth(self, depth):
        self.__depth = depth


b1 = Box(100, 100, 100)
print(b1)
print(
    "상자의 부피는", b1.getHeight() * b1.getLength() * b1.getDepth()
)
```

### programming 04  
 사각형을 나타내는 Rectangle 클래스를 작성해보자.

```python
class Rectangle:

    def __init__(self, x, y, w, h):
        self.__x = x
        self.__y = y
        self.__width = w
        self.__height = h

    def __str__(self):
        return f"Rectangle(x={self.__x}, y={self.__y}, w={self.__width}, h={self.__height})"

    def getX(self):
        return self.__x

    def setX(self, x):
        self.__x = x

    def getY(self):
        return self.__y

    def setY(self, y):
        self.__y = y

    def getWidth(self):
        return self.__width

    def setWidth(self, w):
        self.__width = w

    def getHeight(self):
        return self.__height

    def setHeight(self, h):
        self.__height = h

    def getArea(self):
        return self.__width * self.__height

    def overlap(self, r):
        # 겹치지 않는 4가지 조건 확인
        if (
            self.__x + self.__width <= r.getX()
            or r.getX() + r.getWidth() <= self.__x
            or self.__y + self.__height <= r.getY()
            or r.getY() + r.getHeight() <= self.__y
        ):
            return False  # 겹치지 않음
        return True  # 겹침


# 테스트 코드
r1 = Rectangle(0, 0, 100, 100)
r2 = Rectangle(10, 10, 100, 100)

if r1.overlap(r2):
    print("r1과 r2는 서로 겹칩니다.")
else:
    print("r1과 r2는 서로 겹치지 않습니다.")
```

### programming 05  
삼각형을 나타내는 클래스  Triangle을 작성해보자. 

```python
class Triangle:

    def __init__(self, a1, a2, a3):
        self.__angle1 = a1
        self.__angle2 = a2
        self.__angle3 = a3
        self.numberOfSides = 3  # 기본값 3

    def __str__(self):
        return f"Triangle({self.__angle1}, {self.__angle2}, {self.__angle3})"

    # 접근자 및 설정자 메소드들
    def getAngle1(self):
        return self.__angle1

    def setAngle1(self, a):
        self.__angle1 = a

    def getAngle2(self):
        return self.__angle2

    def setAngle2(self, a):
        self.__angle2 = a

    def getAngle3(self):
        return self.__angle3

    def setAngle3(self, a):
        self.__angle3 = a

    # 세 각의 합이 180도인지 검사
    def checkAngles(self):
        return (self.__angle1 + self.__angle2 + self.__angle3) == 180


# 테스트 코드
triangle = Triangle(90, 30, 60)
print(triangle.checkAngles())  # 90 + 30 + 60 = 180이므로 True 출력
```

### programming 06  
Person이라는 클래스를 작성해보자.  

```python
class Person:

    def __init__(self, name, mobile=None, office=None, email=None):
        self.__name = name
        self.__mobile = mobile
        self.__office = office
        self.__email = email

    def __str__(self):
        res = f"{self.__name}\n"
        if self.__mobile:
            res += f"mobile phone: {self.__mobile}\n"
        if self.__office:
            res += f"office phone: {self.__office}\n"
        if self.__email:
            res += f"email address: {self.__email}\n"
        return res.strip()

    def getName(self):
        return self.__name

    def setName(self, name):
        self.__name = name

    def getMobile(self):
        return self.__mobile

    def setMobile(self, mobile):
        self.__mobile = mobile

    def getOffice(self):
        return self.__office

    def setOffice(self, office):
        self.__office = office

    def getEmail(self):
        return self.__email

    def setEmail(self, email):
        self.__email = email


p1 = Person("Kim", office="1234567", email="kim@company.com")
p2 = Person("Park", office="2345678")
p2.setEmail("park@company.com")
```

### programming 07  
사람들의 연락처를 저장하는  PhoneBook 클래스를 작성해보자

```python
class PhoneBook:

    def __init__(self):
        self.contacts = {}

    def __str__(self):
        result = []
        for name, p in self.contacts.items():
            result.append(str(p))
        return "\n\n".join(result)

    def add(
        self,
        name,
        mobile=None,
        office=None,
        email=None,
    ):
        # 06번의 Person 클래스를 이용하여 contacts 딕셔너리에 저장
        self.contacts[name] = Person(name, mobile, office, email)


obj = PhoneBook()
obj.add("Kim", office="1234567", email="kim@company.com")
obj.add("Park", office="2345678", email="park@company.com")
print(obj)
```

### programing 09  
터틀 그래픽에서 각각의 거북이는 객체이다. 2개의 거북이를 생성하여서  다음과같은 서로 다른 방향으로 움직이도록 하자.

```python
import turtle

# 1. 첫 번째 거북이 생성 (t1)
t1 = turtle.Turtle()
t1.shape("circle")  # 모양을 원(circle)으로 설정

# t1 이동 경로: 왼쪽으로 갔다가 아래로 이동
t1.left(180)  # 왼쪽(180도)을 바라봄
t1.forward(100)  # 100만큼 직진
t1.left(90)  # 왼쪽으로 90도 회전 (아래 방향)
t1.forward(100)  # 100만큼 직진
t1.stamp()  # 현재 위치에 원 모양 도장 찍기

# 2. 두 번째 거북이 생성 (t2)
t2 = turtle.Turtle()
t2.shape("turtle")  # 모양을 거북이(turtle)로 설정

# t2 이동 경로: 오른쪽으로 갔다가 아래로 이동
t2.forward(200)  # 오른쪽으로 200만큼 직진
t2.right(90)  # 오른쪽으로 90도 회전 (아래 방향)
t2.forward(50)  # 50만큼 직진

# 창이 바로 닫히지 않도록 유지
turtle.done()
```
